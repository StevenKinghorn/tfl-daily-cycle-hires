# Creating DuckDB Database
Combining all 48 csv's in to a DuckDB database called tfl_cycle_hires.duckdb. 

Using a DuckDB database will be less cpu and memory hungry.

```python
import pandas as pd
import duckdb
```

```python
clean_data = "../data/tfl-cycling-journey-data-2024-2025/clean-2024-2025-csv/*.csv" #all csv files from CLEAN data folder
DB = "../data/tfl-cycling-journey-data-2024-2025/tfl_cycle_hires.duckdb"            #Database location


with duckdb.connect(DB) as con:   #con.close() is not needed with context manager
    con.sql(f"""
        CREATE TABLE IF NOT EXISTS cycle_hires_table AS 
        SELECT * FROM read_csv_auto('{clean_data}', union_by_name=True)
    """)
    
    row_count = con.sql("SELECT count(*) FROM cycle_hires_table").fetchone()[0]
    print(f"Successfully loaded {row_count} rows from 48 files.\n")

    # Can use: wc -l *.csv command in Linux to count the number of rows in all these csv files.
    # Then just subtract 48 since we do not want to count the headers.


    
```
